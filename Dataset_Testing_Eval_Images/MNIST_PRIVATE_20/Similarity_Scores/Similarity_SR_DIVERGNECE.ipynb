```python
import torch
from torchvision.datasets import MNIST, CIFAR10, USPS, QMNIST, SVHN, FashionMNIST, KMNIST, SEMEION, STL10
import numpy as np
import torch.nn.functional as F
from torch.linalg import matrix_norm
from torchvision import transforms
from math import log
```

```python
#SR_DISTANCE
print("SR_DISTANCE")

mnist = MNIST(root='data', train=True, download=True)
cifar = CIFAR10(root='data', train=True, download=True)
usps = USPS(root='data', train=True, download=True)

data_np_mnist = mnist.data[:5000].numpy().astype(np.float32) / 255.0  # Use subset for speed
data_np_mnist = data_np_mnist[:, None, :, :]
mnist_tensor = torch.from_numpy(data_np_mnist).float()
mnist_tensor = F.interpolate(mnist_tensor, size=(32, 32), mode='bilinear', align_corners=False)
mnist_tensor = mnist_tensor.repeat(1, 3, 1, 1)
data_torch_mnist = mnist_tensor.detach().cpu()
data_torch_mnist_resized = F.interpolate(data_torch_mnist, size=(64, 64), mode='bilinear', align_corners=False)
data_torch_mnist_flat = data_torch_mnist_resized.view(5000, -1)
data_flattened_mnist = data_torch_mnist_flat.flatten()
final_mnist = data_flattened_mnist[:3072*3072]
data_np_mnist = final_mnist.view(3072, 3072).numpy()

data_np_cifar = cifar.data[:5000].astype(np.float32) / 255.0  # Use subset for speed
data_torch_cifar = torch.from_numpy(data_np_cifar.transpose(0, 3, 1, 2))
data_torch_cifar_resized = F.interpolate(data_torch_cifar, size=(64, 64), mode='bilinear', align_corners=False)
data_torch_cifar_flat = data_torch_cifar_resized.view(5000, -1)
data_flattened_cifar = data_torch_cifar_flat.flatten()
final_cifar = data_flattened_cifar[:3072*3072]
data_np_cifar = final_cifar.view(3072, 3072).numpy()

usps = USPS(root='data', train=True, download=True)
data_np_usps = usps.data.astype(np.float32) / 255.0
usps_tensor = torch.from_numpy(data_np_usps[:, None, :, :])
usps_tensor = F.interpolate(usps_tensor, size=(32, 32), mode='bilinear', align_corners=False)
usps_tensor = usps_tensor.repeat(1, 3, 1, 1)
usps_tensor = usps_tensor.view(7291, -1)
data_flattened_usps = usps_tensor.flatten()
target_size = 3072 * 3072
if data_flattened_usps.numel() < target_size:
    data_flattened_usps = F.pad(data_flattened_usps, (0, target_size - data_flattened_usps.numel()))
else:
    data_flattened_usps = data_flattened_usps[:target_size]
final_usps = data_flattened_usps.view(3072, 3072)
data_np_usps = final_usps.cpu().numpy()

qmnist = QMNIST(root='data', train=True, download=True)
data_np_qmnist = qmnist.data.numpy().astype(np.float32) / 255.0
data_np_qmnist = data_np_qmnist[:, None, :, :]  
qmnist_tensor = torch.from_numpy(data_np_qmnist)
qmnist_tensor = F.interpolate(qmnist_tensor, size=(32, 32), mode='bilinear', align_corners=False)
qmnist_tensor = qmnist_tensor.repeat(1, 3, 1, 1)[:5000]
data_torch_qmnist_flat = qmnist_tensor.view(5000, -1)
data_flattened_qmnist = data_torch_qmnist_flat.flatten()
final_qmnist = data_flattened_qmnist[:3072*3072]
data_np_qmnist = final_qmnist.view(3072, 3072).numpy()

svhn_dataset_new = SVHN(root='./data', split="train", download=True)
data_np_svhn = svhn_dataset_new.data.astype(np.float32) / 255.0
svhn_tensor = torch.from_numpy(data_np_svhn[:5000])
svhn_tensor = F.interpolate(svhn_tensor, size=(32, 32), mode='bilinear', align_corners=False)
data_torch_svhn_flat = svhn_tensor.view(5000, -1)
data_flattened_svhn = data_torch_svhn_flat.flatten()
final_svhn = data_flattened_svhn[:3072*3072]
data_np_svhn = final_svhn.view(3072, 3072).numpy()

fashion_fresh = FashionMNIST(root='data', train=True, download=True)
data_np_fashion = fashion_fresh.data.numpy().astype(np.float32) / 255.0
data_np_fashion = data_np_fashion[:, None, :, :]
fashion_tensor = torch.from_numpy(data_np_fashion).float()  
fashion_tensor = F.interpolate(fashion_tensor, size=(32, 32), mode='bilinear', align_corners=False)
fashion_tensor = fashion_tensor.repeat(1, 3, 1, 1) 
fashion_tensor = fashion_tensor[:5000]
fashion_tensor = F.interpolate(fashion_tensor, size=(64, 64), mode='bilinear', align_corners=False)
data_torch_fashion_flat = fashion_tensor.view(5000, -1)
data_flattened_fashion = data_torch_fashion_flat.flatten()
final_fashion = data_flattened_fashion[:3072 * 3072]
data_np_fashion = final_fashion.view(3072, 3072).numpy()

kmnist_fresh = KMNIST(root='data', train=True, download=True)
data_np_kmnist = kmnist_fresh.data.numpy().astype(np.float32) / 255.0
data_np_kmnist = data_np_kmnist[:, None, :, :]
kmnist_tensor = torch.from_numpy(data_np_kmnist).float()  
kmnist_tensor = F.interpolate(kmnist_tensor, size=(32, 32), mode='bilinear', align_corners=False)
kmnist_tensor = kmnist_tensor.repeat(1, 3, 1, 1) 
kmnist_tensor = kmnist_tensor[:5000]
kmnist_tensor = F.interpolate(kmnist_tensor, size=(64, 64), mode='bilinear', align_corners=False)
data_torch_kmnist_flat = kmnist_tensor.view(5000, -1)
data_flattened_kmnist = data_torch_kmnist_flat.flatten()
final_kmnist = data_flattened_kmnist[:3072 * 3072]
data_np_kmnist = final_kmnist.view(3072, 3072).numpy()

transform = transforms.Compose([
    transforms.ToTensor()
])
semeion = SEMEION(root='data', download=True, transform=transform)
semeion_data_raw = []
for i in range(len(semeion)):
    img, _ = semeion[i]
    semeion_data_raw.append(img.squeeze().numpy())
data_np_semeion = np.array(semeion_data_raw).astype(np.float32)
data_np_semeion = data_np_semeion[:, None, :, :]
semeion_tensor = torch.from_numpy(data_np_semeion).float()
semeion_tensor = F.interpolate(semeion_tensor, size=(32, 32), mode='bilinear', align_corners=False) 
semeion_tensor = semeion_tensor.repeat(1, 3, 1, 1)
semeion_tensor = F.interpolate(semeion_tensor, size=(64, 64), mode='bilinear', align_corners=False)
semeion_tensor = semeion_tensor.view(len(semeion), -1) 
data_flattened_semeion = semeion_tensor.flatten()
target_size = 3072 * 3072
if data_flattened_semeion.numel() < target_size:
    data_flattened_semeion = F.pad(data_flattened_semeion, (0, target_size - data_flattened_semeion.numel()))
else:
    data_flattened_semeion = data_flattened_semeion[:target_size]
final_semeion = data_flattened_semeion.view(3072, 3072)
data_np_semeion = final_semeion.numpy()

stl_fresh = STL10(root='./data', split="train", download=True)
stl_data_list = []
for i in range(min(5000, len(stl_fresh))):
    img, _ = stl_fresh[i]
    img_np = np.array(img)  
    if img_np.shape[0] == 3:
        img_np = img_np.transpose(1, 2, 0)
    img_np = (img_np * 255).astype(np.uint8)
    stl_data_list.append(img_np)
data_np_stl = np.array(stl_data_list).astype(np.float32) / 255.0
data_np_stl = data_np_stl.transpose(0, 3, 1, 2)
stl_tensor = torch.from_numpy(data_np_stl).float()
stl_tensor = F.interpolate(stl_tensor, size=(64, 64), mode='bilinear', align_corners=False)
stl_tensor = stl_tensor.reshape(5000, -1)
data_flattened_stl = stl_tensor.flatten()
final_stl = data_flattened_stl[:3072 * 3072]  
data_np_stl = final_stl.view(3072, 3072).numpy()



print(data_np_cifar.shape, data_np_mnist.shape, data_np_usps.shape, data_np_qmnist.shape, data_np_svhn.shape, data_np_fashion.shape, data_np_kmnist.shape, data_np_semeion.shape, data_np_stl.shape)
```

```python
data_np_cifar_inv = np.linalg.pinv(data_np_cifar)
data_np_usps_inv = np.linalg.pinv(data_np_usps)
data_np_qmnist_inv = np.linalg.pinv(data_np_qmnist)
data_np_svhn_inv = np.linalg.pinv(data_np_svhn)
data_np_fashion_inv = np.linalg.pinv(data_np_fashion)
data_np_kmnist_inv = np.linalg.pinv(data_np_kmnist)
data_np_semeion_inv = np.linalg.pinv(data_np_semeion)
data_np_stl_inv = np.linalg.pinv(data_np_stl)

print(data_np_cifar_inv.shape, data_np_usps_inv.shape, data_np_qmnist_inv.shape, data_np_svhn_inv.shape, data_np_fashion_inv.shape, data_np_kmnist_inv.shape, data_np_semeion_inv.shape, data_np_stl_inv.shape)

transformation_matrix_cifar = data_np_mnist @ data_np_cifar_inv
tranformation_matrix_usps = data_np_mnist @ data_np_usps_inv
transformation_matrix_qmnist = data_np_mnist @ data_np_qmnist_inv
transformation_matrix_svhn = data_np_mnist @ data_np_svhn_inv
transformation_matrix_fashion = data_np_mnist @ data_np_fashion_inv
transformation_matrix_kmnist = data_np_mnist @ data_np_kmnist_inv
transformation_matrix_semeion = data_np_mnist @ data_np_semeion_inv
transformation_matrix_stl = data_np_mnist @ data_np_stl_inv
print(transformation_matrix_cifar)
print(tranformation_matrix_usps)

I_cifar = np.eye(3072)
I_usps = np.eye(3072)
I_qmnist = np.eye(3072)
I_svhn = np.eye(3072)
I_fashion = np.eye(3072)
I_kmnist = np.eye(3072)
I_semeion = np.eye(3072)
I_stl = np.eye(3072)

norm_bef_cifar = transformation_matrix_cifar - I_cifar
norm_bef_usps = tranformation_matrix_usps - I_usps
norm_bef_qmnist = transformation_matrix_qmnist - I_qmnist
norm_bef_svhn = transformation_matrix_svhn - I_svhn
norm_bef_fashion = transformation_matrix_fashion - I_fashion
norm_bef_kmnist = transformation_matrix_kmnist - I_kmnist
norm_bef_semeion = transformation_matrix_semeion - I_semeion
norm_bef_stl = transformation_matrix_stl - I_stl

norm_af_cifar = np.linalg.matrix_norm(norm_bef_cifar, ord=-1)
nor_af_usps = np.linalg.matrix_norm(norm_bef_usps, ord=-1)
norm_af_qmnist = np.linalg.matrix_norm(norm_bef_qmnist, ord=-1)
norm_af_svhn = np.linalg.matrix_norm(norm_bef_svhn, ord=-1)
norm_af_fashion = np.linalg.matrix_norm(norm_bef_fashion, ord=-1)
norm_af_kmnist = np.linalg.matrix_norm(norm_bef_kmnist, ord=-1)
norm_af_semeion = np.linalg.matrix_norm(norm_bef_semeion, ord=-1)
norm_af_stl = np.linalg.matrix_norm(norm_bef_stl, ord=-1)

print("CIFAR:", log(norm_af_cifar))
print("USPS:", log(nor_af_usps))
print("QM-NIST:", log(norm_af_qmnist))
print("SVHN:", log(norm_af_svhn))
print("Fashion-MNIST:", log(norm_af_fashion))
print("KMNIST:", log(norm_af_kmnist))
print("SEMEION:", log(norm_af_semeion))
print("STL-10:", log(norm_af_stl))
```
